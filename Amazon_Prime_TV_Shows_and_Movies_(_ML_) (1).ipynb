## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# readyWrite your code to make your dataset analysis.
# ==========================================
# Data Wrangling Code
# ==========================================

# Create copies of original datasets
titles_df = titles.copy()
credits_df = credits.copy()

# ------------------------------------------
# 1. Remove Duplicate Records
# ------------------------------------------
titles_df.drop_duplicates(inplace=True)
credits_df.drop_duplicates(inplace=True)

# ------------------------------------------
# 2. Check Missing Values
# ------------------------------------------
print(titles_df.isnull().sum())
print(credits_df.isnull().sum())

# ------------------------------------------
# 3. Handle Missing Values
# ------------------------------------------
titles_df['age_certification'].fillna('Unknown', inplace=True)
titles_df['runtime'].fillna(titles_df['runtime'].median(), inplace=True)
titles_df['imdb_score'].fillna(titles_df['imdb_score'].median(), inplace=True)
titles_df['tmdb_score'].fillna(titles_df['tmdb_score'].median(), inplace=True)
titles_df['tmdb_popularity'].fillna(titles_df['tmdb_popularity'].median(), inplace=True)
titles_df['seasons'].fillna(0, inplace=True)

# ------------------------------------------
# 4. Convert Data Types
# ------------------------------------------
titles_df['release_year'] = titles_df['release_year'].astype(int)
titles_df['runtime'] = titles_df['runtime'].astype(int)

# ------------------------------------------
# 5. Merge Titles and Credits Dataset
# ------------------------------------------
amazon_df = pd.merge(
    titles_df,
    credits_df,
    on='id',
    how='left'
)

# ------------------------------------------
# 6. Check Final Dataset Shape
# ------------------------------------------
print("Final Dataset Shape :", amazon_df.shape)

# ------------------------------------------
# 7. Check Remaining Missing Values
# ------------------------------------------
print(amazon_df.isnull().sum())

# ------------------------------------------
# 8. Display Final Dataset
# ------------------------------------------
amazon_df.head()

# ------------------------------------------
# 9. Save Cleaned Dataset
# ------------------------------------------
amazon_df.to_csv('AmazonPrime_Cleaned.csv', index=False)
````

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import ttest_ind

movies = titles[titles['type'] == 'MOVIE']['imdb_score'].dropna()
tvshows = titles[titles['type'] == 'SHOW']['imdb_score'].dropna()

t_stat, p_value = ttest_ind(movies, tvshows)

print("T-Statistic :", t_stat)
print("P-Value :", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

Independent Two Sample T-Test

##### Why did you choose the specific statistical test?

The Independent T-Test is appropriate because it compares the mean IMDb scores of two independent groups (Movies and TV Shows) to determine whether their average ratings differ significantly.

#### 2. Perform an appropriate statistical test.

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test
